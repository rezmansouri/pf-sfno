```python
import numpy as np
import matplotlib.pyplot as plt
from utils import get_cr_dirs, collect_sim_paths, HUXDataset
```

```python
np_loaded = np.load("test_0_predictions.npz")
hux = np_loaded["predictions"]
mas = np_loaded["targets"]
```

```python
for i in range(0, hux.shape[0], 5):
    vmin = min(np.min(hux), np.min(mas))
    vmax = max(np.max(hux), np.max(mas))
    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    plt.imshow(hux[i, :, :], cmap='gnuplot', vmin=vmin, vmax=vmax)
    plt.title(f"HUX {i}")
    plt.subplot(1, 2, 2)
    plt.imshow(mas[i, :, :], cmap='gnuplot', vmin=vmin, vmax=vmax)
    plt.title(f"MAS {i}")
    plt.colorbar()
    plt.show()
```
